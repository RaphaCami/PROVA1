**Menor de 16 anos com saldo sufiente**

```python
idade = int(input("Qual é sua idade?: "))
dinheiro = float(input("Quanto seu saldo na carteira? R$ "))


if idade < 16:
    print("Entrada permitida apenas com acompanhante.")
elif dinheiro >= 30:
    print("Ingresso comprado! Bom filme.")
else:
    print("Saldo insuficiente para o ingresso.")

print("Foi bom te atender!")
```
